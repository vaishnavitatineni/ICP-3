```python
import pandas as pd

#loading dataset
df = pd.read_csv('https://raw.githubusercontent.com/JuliaData/CSV.jl/refs/heads/main/test/testfiles/Sacramentorealestatetransactions.csv')

df.head()
```

```python
#question 1
num_columns = len(df.columns)
print("Number of columns:", num_columns)

#question 2
mean_price = df['price'].mean()
print("Mean house price:", mean_price)

#question 3
median_price = df['price'].median()
print("Median house price:", median_price)

#question 4
different_types = df['type'].unique()
print("Different property types:", different_types)

#question 5
max_condo_price = df[df['type'].str.lower() == 'condo']['price'].max()
print("Maximum price of a condo:", max_condo_price)

#question 6
mean_price_by_type = df.groupby('type')['price'].mean()
print("Mean price by property type:\n", mean_price_by_type)

#question 7
count_by_type = df['type'].value_counts()
print("Count of each property type:\n", count_by_type)
```
